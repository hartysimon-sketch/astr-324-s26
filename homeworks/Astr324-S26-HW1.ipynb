## Problem 1
* Generate a sample of 10,000 data values drawn from N($\mu$=1.0, $\sigma$=0.2)and draw a pretty histogram
with the bin size determined using the Freedman-Diaconis rule. Show (Poissonian) uncertainty ($N_{bin} \pm \sqrt{N_{bin}}$, where $N_{bin}$ is the count in a given bin) for each histogram value, and overplot the true distribution.

```python
np.random.seed(42)
mu, sigma, numpts = 1, 0.2, 10000
data = stats.norm(loc=mu, scale=sigma).rvs(numpts)
norm_mean = np.mean(data)

hist, edges = np.histogram(data, bins='fd')
centers = (edges[:-1] + edges[1:]) / 2
bin_width = edges[1] - edges[0]

true_x = np.linspace(np.max(data), np.min(data), 100)
true_y = stats.norm.pdf(true_x, mu, sigma) * numpts * bin_width

plt.plot(true_x, true_y, ls='-', color='purple', label='True Distribution')
plt.hist(data, bins=edges, histtype='stepfilled');
plt.errorbar(centers, hist, label="Poisson Uncertainty", yerr=np.sqrt(hist), 
             capsize=2, color='black', fmt='none', alpha=0.7)

plt.legend()
plt.title(f'10000 Sample Normal Distribution Histogram, mu={mu} sigma ={sigma}')
plt.xlabel('Value')
plt.ylabel('Frequency')

plt.show()
```

## Problem 2
* Do the same for a Cauchy distribution with the same $\mu$ and $\gamma$=0.2.

```python
np.random.seed(42)
mu, gamma, numpts = 1, 0.2, 10000
data = stats.cauchy(loc=mu, scale=gamma).rvs(numpts)
cauchy_mean = np.mean(data)

plot_range = (mu - 10*gamma, mu + 10*gamma)

hist, edges = np.histogram(data, bins='fd', range=plot_range)
centers = (edges[:-1] + edges[1:]) / 2
bin_width = edges[1] - edges[0]

true_x = np.linspace(plot_range[0], plot_range[1], 100)
true_y = stats.cauchy.pdf(true_x, mu, gamma) * numpts * bin_width

plt.plot(true_x, true_y, ls='-', color='purple', label='True Distribution')
plt.errorbar(centers, hist, label="Poisson Uncertainty", yerr=np.sqrt(hist), 
             capsize=2, color='black', fmt='none', alpha=0.7)
plt.hist(data, bins=edges, histtype='stepfilled', range=plot_range);

plt.legend()
plt.title(f'10000 Sample Cauchy Distribution Histogram, mu={mu} gamma ={gamma}')
plt.xlabel('Value')
plt.ylabel('Frequency')

plt.show()
```

## Problem 5:
* Generate a sample of 100 data values drawn from N($\mu$=1.0, $\sigma$=0.1). Draw 3 values from N($\mu$=10.0, $\sigma$=1.0) and add them to the first sample. Compute the mean and median values for this extended sample 
and discuss how similar they are. 

```python
np.random.seed(42)

mu, sigma, numpts = 1, 0.1, 100
data100 = stats.norm(loc=mu, scale=sigma).rvs(numpts)
mean100 = np.mean(data100)
median100 = np.median(data100)

mu, sigma, numpts = 10, 1, 3
data3 = stats.norm(loc=mu, scale=sigma).rvs(numpts)
mean3 = np.mean(data3)
median3 = np.median(data3)

data = np.concatenate((data100, data3))
bigmean = np.mean(data)
bigmedian = np.median(data)

print(f'100 sample mean: {mean100}, 3 sample mean: {mean3}, combined mean: {bigmean}',
      f'100 sample median: {median100}, 3 sample median: {median3}, combined median: {bigmedian}', 
      sep="\n")
```

While both the median and the mean notably change, the mean changes significantly more once the second sample is added. As the 3 values from the second sample are much larger that the values in the initial sample, this makes sense. Because computing the mean directly uses every value in the sample, it is more susceptible to outliers. The median is only slightly shifted because it is the 'middle' value and only 3 outliers were added. The difference of ~ 0.2 between the combined mean and median values is due to this shifting of the mean.

## Problem 7:
* Now there are 3 doors: one with a car, two with goats. The game show host asks you to pick a door, but not to open it yet. Then the host opens one of the other two doors (that you did not pick), making sure to select one with a goat. The host offers you the opportunity to switch doors. Do you switch? Hint: use conditional probabilities.

Yes, you switch. When your guess was initially made, you had a 1/3 chance of selecting the door with a car behind it - this is completely independent of the actual object behind that door, as discussed above. This does not change when the host reveals one of the goats. If I open door one, and the host opens door 2, Using Bayes' Rule we have:
$$p(C_3|H_2) = \frac{p(H_2|C_3)p(C_3)}{p(H_2)} = \frac{1\times\frac{1}{3}}{\frac{1}{2}} = \frac{2}{3}$$
where the denominator $P(H_2)$ can be found using the law of total probabilities:

If the car is at door 1: Host can open doors 2 or 3 $P(H_2 | C_1) \times P(C_1)= 1/6$

If the car is at door 2: Host opens door 3 (cannot open your door and can't reveal car) $P(H_2 | C_2) \times P(C_2) = 0$

If the car is at door 3: Host opens door 2 $P(H_2 | C_3) \times P(C_3) = 1/3$ 

1/6 + 1/3 + 0 = 1/2

So the probability that the car is behind the third door (the door that you didn't choose and the host didn't open) is 2/3. This makes sense as the probability of being wrong on the first guess was 2/3, and after switching you will be correct unless your initial guess was already correct. Because you know that one of the doors has the car behind it, the probability must add to one.


## Problem 8:
* Generate a sample of N=10 data values drawn from $N$($\mu$=1.0, $\sigma$=0.1). Plot the likelihood function for $\mu$ given these 10 values. Overplot the prediction that says it should look like $N$($\overline{x}, \sigma$),
where
$$\overline{x} \equiv \frac{1}{N}\sum_{i=1}^N x_i,$$
and
$\sigma = s/\sqrt{N},$
and comment. 

```python
np.random.seed(42)

#get sample and calculate mean, std, sigma
mu, sigma, numpts = 1, 0.1, 10
data = stats.norm(loc=mu, scale=sigma).rvs(numpts)
mymean = np.mean(data)
sigma_xbar = sigma / np.sqrt(numpts)

#calculate likelihoods of data points
muGrid = np.linspace(0,2,1000)
likelihoods = []
for i, value in enumerate(data):
    temp = stats.norm(value, sigma).pdf(muGrid)
    likelihoods.append(temp)

#calculate likelihood funtion and normalize
L = np.prod(likelihoods, axis=0) 
L /= np.trapezoid(L, muGrid)

#make prediction
prediction = stats.norm(loc=mymean, scale=sigma_xbar).pdf(muGrid)

# plot
plt.plot(muGrid, L, ls='-', c='black', label=r'Likelihood from data')
plt.plot(muGrid, prediction, label='Prediction', ls='--')

plt.xlim(0.2, 1.8)
plt.xlabel('$\mu$')
plt.title('Likelihood for 10 data points drawn from a Gaussian Distribution')
plt.legend()
plt.show()

print(f'Mean = {mymean}')
```

As expected the likelihood and prediction overlap very closely, and the peak of the distribution is the sample mean. This shows how the joint likelihood of multiple gaussians is another gaussian, demonstrating how our uncertainty in the mean goes down as we add more data points.

100 galaxies and found that 30 harbor black holes? And 1000 galaxies, of which 300 harbor black holes. 

Hint: use the binomial likelihood example from Week2 lecture. 

```python
def get_prob(n, k):
    grid = np.linspace(0,1,1000)
    likelihood = stats.binom.pmf(k, n, grid)
    likelihood /= np.trapezoid(likelihood, grid)
    myprob = np.trapezoid(likelihood[grid <= 0.5], grid[grid <= 0.5])

    return myprob
```

```python
params = [(10, 3), (100, 30), (1000, 300)]
probs = []

for n, k in params:
    temp = get_prob(n, k)
    print(f'{n} galaxies probability: {temp}')
```
